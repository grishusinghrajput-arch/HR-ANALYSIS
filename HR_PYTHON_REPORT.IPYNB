```python
df.columns = df.columns.str.strip()

df.columns = df.columns.str.replace(" ", "_")
```

```python
df["Joining_Date"] = pd.to_datetime(
    df["Joining_Date"],
    format="mixed",
    errors="coerce"
)
```

```python
df["Joining_Date"].dtype
```

```python
df["Age"] = df["Age"].astype("int64")

df["Salary"] = df["Salary"].astype("int64")

df["Overtime_Hours"] = df["Overtime_Hours"].astype("int64")

df["Leave_Days"] = df["Leave_Days"].astype("int64")

df["Training_Hours"] = df["Training_Hours"].astype("int64")
```

```python
df["Gender"] = df["Gender"].str.title()

df["Department"] = df["Department"].str.title()

df["Education"] = df["Education"].str.title()

df["Location"] = df["Location"].str.title()

df["Work_Mode"] = df["Work_Mode"].str.title()

df["Attrition"] = df["Attrition"].str.title()

df["Employee_Status"] = df["Employee_Status"].str.title()
```

```python
print(df["Age"].min(), df["Age"].max())

print(df["Salary"].min(), df["Salary"].max())

print(df["Performance_Rating"].unique())

print(df["Job_Satisfaction"].unique())

print(df["Attrition"].unique())
```

```python
print("Total Rows:", df.shape[0])

print("Total Columns:", df.shape[1])

print("Missing Values:", df.isnull().sum().sum())

print("Duplicate Records:", df.duplicated().sum())

print("Duplicate Employee IDs:", df["Employee_ID"].duplicated().sum())

print("Data Types:")
print(df.dtypes)
```

```python
df.to_csv("HR_Analytics_Cleaned.csv", index=False)
```
